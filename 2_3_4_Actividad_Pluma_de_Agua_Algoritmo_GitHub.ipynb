> **¿Por qué los costeños le decimos “pluma” a la llave del agua?**  
>  
> En el Caribe colombiano, abrir la “pluma” no es solo una acción cotidiana: es una expresión cargada de historia, técnica y cultura. Aunque algunos se sorprendan o incluso se burlen del término, lo cierto es que decir “pluma” en vez de “llave” no es un error, sino una herencia viva que conecta nuestro lenguaje con siglos de evolución hidráulica.  
>  
> Mucho antes de que existieran los grifos modernos, el agua se distribuía mediante sistemas regulados con precisión. En España, por ejemplo, se usaba la **“pluma de agua”** como unidad de medida para calcular el caudal que se entregaba a cada hogar. Aproximadamente, una pluma equivalía a **0.025 litros por segundo**. En algunos lugares, ese flujo se regulaba mediante orificios calibrados y dispositivos llamados **plumeros**, que permitían controlar la salida del agua.  
>  
> Durante la época colonial, muchas de estas prácticas técnicas cruzaron el Atlántico y llegaron a ciudades como Cartagena, Barranquilla y Santa Marta. Con el tiempo, el término “pluma” dejó de ser solamente técnico y pasó al habla cotidiana como sinónimo de la llave del agua.

````python
"""
Actividad Integradora: La Pluma de Agua
Autores: Jesus Martinez, Danna Castillo, Jesus Builes, Raul Robles y Lucía Rojas
Curso: Programación Aplicada a la Ingeniería
Fecha: Mayo 2026
Descripción: Algoritmo para modelar el consumo hídrico residencial a partir de
             la unidad de medida histórica "pluma de agua" en el Caribe colombiano.
             Cumple con las directrices de estilo PEP 8.
"""

def clasificar_consumo(consumo_diario: float) -> str:
    """
    Clasifica el nivel de consumo diario de agua según los rangos establecidos.
    """
    if consumo_diario < 1000:
        return "Bajo"
    elif 1000 <= consumo_diario <= 3000:
        return "Medio"
    else:
        return "Alto"


def main():
    # 1. Solicitar al usuario el número de plumas con validación de datos
    try:
        plumas_asignadas = float(input("Ingrese el número de plumas de agua asignadas al hogar: "))
        if plumas_asignadas < 0:
            print("Error: El número de plumas no puede ser un valor negativo.")
            return
    except ValueError:
        print("Error: Por favor, ingrese un número válido (entero o decimal).")
        return

    # 2. Definir constantes del problema (PEP 8: Mayúsculas fijas)
    CAUDAL_POR_PLUMA = 0.025  # Equivalencia física: 1 pluma = 0.025 litros/segundo
    SEGUNDOS_POR_DIA = 86400  # Factor de tiempo: 24 horas * 60 minutos * 60 segundos
    DIAS_DEL_MES = 30         # Periodo estándar para el cálculo mensual de consumo
    LITROS_A_M3 = 1000        # Factor de conversión volumétrica: 1000 litros = 1 m³

    # 3. Calcular el caudal total (Litros por segundo)
    caudal_total_lps = plumas_asignadas * CAUDAL_POR_PLUMA

    # 4. Calcular el consumo diario (Litros y Metros Cúbicos)
    consumo_diario_litros = caudal_total_lps * SEGUNDOS_POR_DIA
    consumo_diario_m3 = consumo_diario_litros / LITROS_A_M3

    # 5. Calcular el consumo mensual (Litros y Metros Cúbicos)
    consumo_mensual_litros = consumo_diario_litros * DIAS_DEL_MES
    consumo_mensual_m3 = consumo_mensual_litros / LITROS_A_M3

    # 6. Clasificar el nivel de consumo según los rangos establecidos
    nivel_consumo = clasificar_consumo(consumo_diario_litros)

    # 7. Mostrar resultados estructurados (Salidas del sistema)
    print("\n" + "="*55)
    print("        REPORTE TÉCNICO: BALANCE DE CONSUMO HÍDRICO")
    print("="*55)
    print(f" Cantidad de plumas asignadas : {plumas_asignadas:.2f}")
    print(f" Caudal total de entrada       : {caudal_total_lps:.4f} L/s")
    print("-"*55)
    print(f" Consumo Diario               : {consumo_diario_litros:,.2f} L  | ({consumo_diario_m3:.3f} m³)")
    print(f" Consumo Mensual (30 días)    : {consumo_mensual_litros:,.2f} L  | ({consumo_mensual_m3:.3f} m³)")
    print(f" Clasificación de Consumo      : {nivel_consumo}")
    print("="*55)


if __name__ == "__main__":
    main()
````

````python
"""
Actividad Integradora: La Pluma de Agua (Con Extensión Opcional)
Autores: Jesus Martinez, Danna Castillo, Jesus Builes, Raul Robles y Lucía Rojas
Curso: Programación Aplicada a la Ingeniería
Fecha: Mayo 2026
Descripción: Algoritmo avanzado que modela el consumo hídrico, incluye
             conversión automática a metros cúbicos y presentación tabular (Punto 8).
             Cumple con las directrices de estilo PEP 8.
"""

def clasificar_consumo(consumo_diario: float) -> str:
    """Clasifica el nivel de consumo diario de agua."""
    if consumo_diario < 1000:
        return "Bajo"
    elif 1000 <= consumo_diario <= 3000:
        return "Medio"
    else:
        return "Alto"


def main():
    # 1. Entrada de datos con validación
    try:
        plumas_asignadas = float(input("Ingrese el número de plumas de agua asignadas al hogar: "))
        if plumas_asignadas < 0:
            print("Error: El número de plumas no puede ser negativo.")
            return
    except ValueError:
        print("Error: Ingrese un número válido.")
        return

    # 2. Constantes físicas y de tiempo
    CAUDAL_POR_PLUMA = 0.025  # Litros por segundo
    SEGUNDOS_POR_DIA = 86400
    LITROS_A_M3 = 1000        # Factor de conversión fundamental (1 m3 = 1000 L)

    # 3. Cálculos de caudales y consumos (Conversiones del Punto 8)
    caudal_total_lps = plumas_asignadas * CAUDAL_POR_PLUMA

    # Consumos en Litros
    diario_litros = caudal_total_lps * SEGUNDOS_POR_DIA
    semanal_litros = diario_litros * 7
    mensual_litros = diario_litros * 30

    # Consumos en Metros Cúbicos (m³) - Extensión Punto 8
    diario_m3 = diario_litros / LITROS_A_M3
    semanal_m3 = semanal_litros / LITROS_A_M3
    mensual_m3 = mensual_litros / LITROS_A_M3

    nivel_consumo = clasificar_consumo(diario_litros)

    # 7 y 8. Presentación de resultados en formato tabular estructurado
    print("\n" + "="*60)
    print("        REPORTE TÉCNICO DE CONSUMO HÍDRICO RESIDENCIAL")
    print("="*60)
    print(f" Cantidad de plumas evaluadas : {plumas_asignadas:.2f}")
    print(f" Caudal total de entrada       : {caudal_total_lps:.4f} L/s")
    print(f" Clasificación del Consumo     : {nivel_consumo}")
    print("-"*60)
    print(f"{'PERIODO':<15} | {'CONSUMO (Litros)':<18} | {'CONSUMO (m³)':<12}")
    print("-"*60)
    print(f"{'Diario':<15} | {diario_litros:<18,.2f} | {diario_m3:<12.3f}")
    print(f"{'Semanal':<15} | {semanal_litros:<18,.2f} | {semanal_m3:<12.3f}")
    print(f"{'Mensual (30 d)':<15} | {mensual_litros:<18,.2f} | {mensual_m3:<12.3f}")
    print("="*60)


if __name__ == "__main__":
    main()
````
