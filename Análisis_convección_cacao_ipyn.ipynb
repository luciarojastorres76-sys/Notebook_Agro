```
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Nuevos datos de Cacao por Convección Forzada
datos_cacao = {
    'tiempo_secado': [24.1, 25.0, 26.2, 20.3, 21.1, 22.4, 17.2, 17.8, 18.9, 14.8, 16.1, 17.3],
    'velocidad_aire': (['T1_1.0ms']*3 + ['T2_1.5ms']*3 + ['T3_2.0ms']*3 + ['T4_2.5ms']*3),
    'lote_cacao': ['Lote_A', 'Lote_B', 'Lote_C'] * 4
}

df_cacao = pd.DataFrame(datos_cacao)

# Modelo DBCA (Diseño de Bloques Completos al Azar)
# velocidad_aire es el tratamiento y lote_cacao es el bloque
modelo = ols('tiempo_secado ~ velocidad_aire + lote_cacao', data=df_cacao).fit()

# Generar Tabla ANOVA
print("=== Análisis de Varianza - Cacao ===")
tabla_anova = sm.stats.anova_lm(modelo, typ=2)
print(tabla_anova)
```
